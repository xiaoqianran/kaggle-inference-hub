# 启动

## 第一步 从github拉取wheel

```python
import os
import shutil
import subprocess
from pathlib import Path

# ============================================================
# Kaggle / uv 设置
# ============================================================

os.environ["UV_LINK_MODE"] = "copy"
os.environ.pop("UV_SYSTEM_PYTHON", None)

ROOT = Path("/kaggle/working/TripoSR")
REPO = "https://github.com/VAST-AI-Research/TripoSR.git"

UV = "/usr/local/bin/uv"
PY = ROOT / ".venv/bin/python"

WHEEL_URL = (
    "https://github.com/xiaoqianran/kaggle-build/releases/download/"
    "triposr-py310-torch2.7.1-cu128-sm75/"
    "torchmcubes-0.1.0-cp310-cp310-linux_x86_64.whl"
)


def run(cmd, cwd=None):
    print("\n>>>", " ".join(map(str, cmd)))
    subprocess.run(
        list(map(str, cmd)),
        check=True,
        cwd=str(cwd) if cwd else None,
    )


# ============================================================
# 1. 清理 + Clone TripoSR
# ============================================================

if ROOT.exists():
    shutil.rmtree(ROOT)

run([
    "git",
    "clone",
    "--depth", "1",
    REPO,
    ROOT,
])


# ============================================================
# 2. 创建 Python 3.10 venv
# ============================================================

run([
    UV,
    "venv",
    "--no-managed-python",
    "--python", "/usr/bin/python3.10",
    ROOT / ".venv",
])


# ============================================================
# 3. 安装 PyTorch 2.7.1 + CUDA 12.8
# ============================================================

run([
    UV,
    "pip",
    "install",
    "--python", PY,

    "torch==2.7.1",
    "torchvision==0.22.1",

    "--index-url",
    "https://download.pytorch.org/whl/cu128",
])


# ============================================================
# 4. 去掉 requirements 里的 torchmcubes
#
# torchmcubes 使用我们自己已经编译好的 Release wheel
# rembg 保留
# ============================================================

REQ = ROOT / "requirements.txt"
REQ_FIXED = ROOT / "requirements.kaggle.txt"

lines = [
    line
    for line in REQ.read_text().splitlines()
    if "torchmcubes" not in line.lower()
]

REQ_FIXED.write_text(
    "\n".join(lines) + "\n"
)

print("\nGenerated:", REQ_FIXED)


# ============================================================
# 5. 安装 TripoSR 其他依赖
# ============================================================

run([
    UV,
    "pip",
    "install",
    "--python", PY,
    "-r", REQ_FIXED,
])


# ============================================================
# 6. 给 rembg 增加 GPU backend
#
# rembg[gpu] -> onnxruntime-gpu
# ============================================================

run([
    UV,
    "pip",
    "install",
    "--python", PY,
    "rembg[gpu]",
])


# ============================================================
# 7. 安装预编译 torchmcubes
#
# 不再 CMake / nvcc 编译
# ============================================================

run([
    UV,
    "pip",
    "install",
    "--python", PY,
    WHEEL_URL,
])


# ============================================================
# 8. 基础验证
# ============================================================

run([
    PY,
    "-c",
    r'''
import torch
import torchmcubes
import onnxruntime as ort

print()
print("========================================")
print("TripoSR Kaggle Environment")
print("========================================")

print("Python environment OK")
print()

print("Torch:", torch.__version__)
print("Torch CUDA:", torch.version.cuda)
print("CUDA available:", torch.cuda.is_available())

assert torch.cuda.is_available(), "PyTorch CUDA 不可用"

print("GPU:", torch.cuda.get_device_name(0))
print("Capability:", torch.cuda.get_device_capability(0))

print()
print("ONNX Runtime:", ort.__version__)
print("ORT providers:", ort.get_available_providers())

assert "CUDAExecutionProvider" in ort.get_available_providers(), (
    "onnxruntime-gpu 已安装，但 CUDAExecutionProvider 不可用"
)

print()

# 实际运行 torchmcubes CUDA kernel
x = torch.randn(
    16, 16, 16,
    device="cuda"
)

v, f = torchmcubes.marching_cubes(
    x,
    0.0
)

print("torchmcubes CUDA OK")
print("vertices:", v.shape)
print("faces:", f.shape)

print("========================================")
'''
])


# ============================================================
# 9. 验证 rembg 实际 Session 是否使用 GPU
# ============================================================

run([
    PY,
    "-c",
    r'''
import rembg

print()
print("Creating rembg session...")

session = rembg.new_session("u2net")

providers = session.inner_session.get_providers()

print("rembg active providers:", providers)

assert "CUDAExecutionProvider" in providers, (
    "rembg Session 没有使用 CUDAExecutionProvider"
)

print("✅ rembg GPU OK")
'''
])


print()
print("========================================")
print("✅ TripoSR installation completed")
print("========================================")
print("Python:", PY)
print("Root:", ROOT)
```
